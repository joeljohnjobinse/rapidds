# rapidds — Canonical Notebook Example

This notebook demonstrates a minimal, end-to-end rapidds workflow.
It is intended as a *starting point* for real analysis, not a full tutorial.

```python
import pandas as pd
from rapidds import Dataset
from rapidds.visualize import plot_missing
```

## Load data
Replace `data.csv` with your actual dataset.

```python
df = pd.read_csv("data.csv")
df.head()
```

## Wrap the dataset with rapidds

```python
ds = Dataset(df)
ds
```

## Inspect data quality

```python
report = ds.inspect()
report.show()
```

## Optional: visualize missing values

```python
plot_missing(ds)
```

## Apply explicit cleaning decisions

```python
df_clean = ds.clean(
    num_missing="median",
    cat_missing="mode"
)
df_clean.head()
```

## Prepare for modeling

```python
X, y, warnings = ds.prepare_for_modeling(target="target_column")
warnings
```
